```python
def define_hosts(host_names):
    hosts = [
        {
            "host": host_name + ".polytechnique.fr",
            "username": "Put your username here",
            "password": "Put your password here",
        }
        for host_name in host_names
    ]
    return hosts
```

```python
import paramiko
import re

def get_gpus_via_ssh(hosts):
    """
    Connects to remote hosts via SSH, runs `nvidia-smi`, and retrieves GPU information.
    
    :param hosts: List of dictionaries, each containing "host", "username", and "password".
    :return: Dictionary mapping hostnames to GPU details.
    """
    gpu_info = {}

    for host in hosts:
        hostname = host["host"]
        username = host["username"]
        password = host["password"]
        
        try:
            # SSH Connection
            ssh = paramiko.SSHClient()
            ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())
            ssh.connect(hostname, username=username, password=password, timeout=10)

            # Run nvidia-smi
            stdin, stdout, stderr = ssh.exec_command("nvidia-smi --query-gpu=name,memory.total --format=csv,noheader")
            gpu_data = stdout.read().decode().strip()

            if gpu_data:
                gpus = [re.split(r",\s*", line) for line in gpu_data.split("\n")]
                gpu_info[hostname] = [{"name": gpu[0], "memory": gpu[1]} for gpu in gpus]
            else:
                gpu_info[hostname] = "No GPU found"

            ssh.close()
        except Exception as e:
            gpu_info[hostname] = f"Error: {str(e)}"

    return gpu_info
```

```python
gpu_details = get_gpus_via_ssh(hosts)

# Print GPU info
for host, gpus in gpu_details.items():
    print(f"\nHost: {host}")
    print(gpus)
```

```python
salles = {
    "salle_30": ["allemagne", "angleterre", "autriche", "belgique", "espagne", "finlande"],
    "salle_31": ["albatros", "autruche", "bengali", "coucou", "dindon", "epervier"],
    "salle_32": ["aerides", "barlia", "calanthe", "diuris", "encyclia", "epipactis"],
    "salle_33": ["ain", "allier", "ardennes", "carmor", "charente", "cher"],
    "salle_34": ["ablette", "anchois", "anguille", "barbeau", "barbue", "baudroie"],
    "salle_35": ["acromion", "apophyse", "astragale", "atlas", "axis", "coccyx"],
    "salle_36": ["bentley", "bugatti", "cadillac", "chrysler", "corvette", "ferrari"],
}
```

```python
for salle, computer_list in salles.items():
    print(f"\nSalle: {salle}")
    hosts = define_hosts(computer_list)
    gpu_details = get_gpus_via_ssh(hosts)
    for host, gpus in gpu_details.items():
        print(f"\nHost: {host}")
        print(gpus)
    
    print("\n" + "-"*50 + "\n")
```
