```python
import os
import pandas as pd

# Path to the full dataset
DATASET_PATH = os.path.expanduser("/users/eleves-b/2022/wajdi.maatouk/NLP_project/french_wikipedia_articles.csv")

# Number of machines / chunks
NUM_CHUNKS = 25

def split_dataset():
    df = pd.read_csv(DATASET_PATH)
    df = df.head(5000)
    total_len = len(df)
    chunk_size = total_len // NUM_CHUNKS

    for i in range(NUM_CHUNKS):
        start_idx = i * chunk_size
        # Make the last chunk extend to the end of the dataset
        end_idx = (i + 1) * chunk_size if i < (NUM_CHUNKS - 1) else total_len

        chunk_df = df.iloc[start_idx:end_idx]
        out_path = os.path.expanduser(f"~/NLP_project/datasets/split_{i}.csv")
        chunk_df.to_csv(out_path, index=False)
        print(f"Created chunk {i}: {out_path}")

split_dataset()
```
